```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

# Her model için 4 görüntü (örnek isimler)
sistnet_imgs1 = ["sist-0854.png", "sist-1386.png", "sist-2250.png", "sist-3930.png"]
sistnet_imgs2 = ["sist-5333.png", "sist-5554.png", "sist-6339.png", "sist-6934.png"]

mobilenet_imgs1 = ["mob-0854.png", "mob-1386.png", "mob-2250.png", "mob-3930.png"]
mobilenet_imgs2 = ["mob-5333.png", "mob-5554.png", "mob-6339.png", "mob-6934.png"]

effdet_imgs1 = ["eff-0854.png", "eff-1386.png", "eff-2250.png", "eff-3930.png"]
effdet_imgs2 = ["eff-5333.png", "eff-5554.png", "eff-6339.png", "eff-6934.png"]

yolo_imgs1 = ["yolo-0854.png", "yolo-1386.png", "yolo-2250.png", "yolo-3930.png"]
yolo_imgs2 = ["yolo-5333.png", "yolo-5554.png", "yolo-6339.png", "yolo-6934.png"]

all_models1 = [sistnet_imgs1, mobilenet_imgs1, effdet_imgs1, yolo_imgs1]
all_models2 = [sistnet_imgs2, mobilenet_imgs2, effdet_imgs2, yolo_imgs2]

titles = [
    "a) SiStNet (mAP 0.059)",
    "b) MobileNetV3 (mAP 0.064)",
    "c) EfficientDet (mAP 0.0)",
    "d) YOLOv5 (mAP 0.093)"
]
```

```python
fig, axes = plt.subplots(4, 4, figsize=(14, 10))

for col in range(4):
    for row in range(4):
        img_name = all_models1[col][row]
        img = mpimg.imread(f'./inference 5 epoch/{img_name}')

        axes[row, col].imshow(img, aspect='auto')
        axes[row, col].set_xticks([])
        axes[row, col].set_yticks([])

        # axis frame kaldır (çok önemli)
        for spine in axes[row, col].spines.values():
            spine.set_visible(False)

        if row == 0:
            axes[row, col].set_title(titles[col], fontsize=12, pad=2)

# FULL sıkıştırma
plt.subplots_adjust(
    wspace=0.03,
    hspace=0.1,
    left=0.0,
    right=1.0,
    top=0.93,
    bottom=0.0
)

plt.show()
```
