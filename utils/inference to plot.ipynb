```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

# Her model için 4 görüntü (örnek isimler)
sistnet_imgs1 = ["s1.png", "s2.png", "s3.png", "s4.png"]
sistnet_imgs2 = ["s5.png", "s6.png", "s7.png", "s8.png"]

mobilenet_imgs1 = ["f1.png", "f2.png", "f3.png", "f4.png"]
mobilenet_imgs2 = ["f5.png", "f6.png", "f7.png", "f8.png"]

effdet_imgs1 = ["e1.png", "e2.png", "e3.png", "e4.png"]
effdet_imgs2 = ["e5.png", "e6.png", "e7.png", "e8.png"]

yolo_imgs1 = ["y1.png", "y2.png", "y3.png", "y4.png"]
yolo_imgs2 = ["y5.png", "y6.png", "y7.png", "y8.png"]

all_models1 = [sistnet_imgs1, mobilenet_imgs1, effdet_imgs1, yolo_imgs1]
all_models2 = [sistnet_imgs2, mobilenet_imgs2, effdet_imgs2, yolo_imgs2]

#titles = [
#    "a) SiStNet (mAP 0.059)",
#    "b) Faster R-CNN (mAP 0.064)",
#    "c) EfficientDet (mAP 0.0)",
#    "d) YOLOv5 (mAP 0.093)"
#]

titles = [
    "a) SiStNet",
    "b) Faster R-CNN",
    "c) EfficientDet",
    "d) YOLOv5"
]
```

```python
fig, axes = plt.subplots(4, 4, figsize=(14, 10))

for col in range(4):
    for row in range(4):
        img_name = all_models1[col][row]
        img = mpimg.imread(f'../../../article_eval_abl_study/preds/{img_name}')

        axes[row, col].imshow(img, aspect='auto')
        axes[row, col].set_xticks([])
        axes[row, col].set_yticks([])

        # axis frame kaldır (çok önemli)
        for spine in axes[row, col].spines.values():
            spine.set_visible(False)

        if row == 0:
            axes[row, col].set_title(titles[col], fontsize=12, pad=2)

# FULL sıkıştırma
plt.subplots_adjust(
    wspace=0.03,
    hspace=0.1,
    left=0.0,
    right=1.0,
    top=0.93,
    bottom=0.0
)

# PNG kaydet
plt.savefig(
    "comparison_grid_1.png",
    dpi=300,
    bbox_inches='tight',
    pad_inches=0
)

plt.show()
```
